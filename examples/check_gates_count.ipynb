```python
import time
import json
import csv

from qiskit import QuantumCircuit
from qiskit.transpiler import CouplingMap, PassManager
from pathlib import Path
from typing import Dict, List, Iterable
from util import EAGLE_COUPLING, sabre, count_swaps 
from multilevel_sabre import MultiLevelSabre
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit.transpiler.passes import VF2Layout
from collections import defaultdict


```

```python
def circuit_info(qc):
    counts = defaultdict(int)
    for inst in qc.data:
        counts[len(inst.qubits)] += 1

    print(f"    Gate counts: {dict(counts)}")
    print(f"        Ops: {qc.count_ops()}")
```

```python
directory = "circuits_transpiled/"
circuit_files = []
for file in Path(directory).glob("*.qasm"):
    circuit_files.append(file)
print(f"Found {len(circuit_files)} circuit files.")    
```

```python
for qasm_file in circuit_files:
    circuit = QuantumCircuit.from_qasm_file(qasm_file)
    print(f"Circuit: {qasm_file.name}")
    circuit_info(circuit)
```
